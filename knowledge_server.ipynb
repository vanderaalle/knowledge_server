# Knowledge Server Notebook
Interface for indexing PDFs and querying the knowledge base via Ollama + Qdrant.

```python
import os

# Environment variables — must be set before importing server
os.environ['OLLAMA_MODEL']    = 'mxbai-embed-large'
os.environ['QDRANT_HOST']     = 'localhost'
os.environ['QDRANT_PORT']     = '6333'
os.environ['COLLECTION_NAME'] = 'pdf_library'
os.environ['TESSERACT_CMD']   = '/usr/bin/tesseract'
os.environ['POPPLER_PATH']    = '/usr/bin'
```

```python
import sys
sys.path.insert(0, '.')

import server
print('Server module loaded OK')
```

## Index a PDF library
Point `PDF_DIR` to the folder containing your PDFs. Run once (or re-run to update).

```python
PDF_DIR = '/path/to/your/pdfs'  # <-- change this

try:
    result = server.index_library(PDF_DIR)
    print(result)
except TypeError:
    # Function may be async
    import asyncio
    result = asyncio.run(server.index_library(PDF_DIR))
    print(result)
```

## Query the library

```python
QUERY = 'spectral music notation'  # <-- change this
N_RESULTS = 5

try:
    results = server.query_library(QUERY, n_results=N_RESULTS)
except TypeError:
    import asyncio
    results = asyncio.run(server.query_library(QUERY, n_results=N_RESULTS))

print(results)
```
